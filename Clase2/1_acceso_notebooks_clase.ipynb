# Creación de repositorio en Github

- Ir al + y "Create New Repository".
- Colocarle un nombre.
- Dejarlo como "Public" y no agregar nada más.
- Dar click en "Create repository".
- Dar click en "Get started by creating a new file" y crear un archivo "prueba.ipynb" o crear una carpeta y un archivo "clase_2/prueba.ipynb".
